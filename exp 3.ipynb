```python
#change a given string to new string
s=input("enter a string:")
newstring=s[-1]+s[1:-1]+s[0]
print("new string:",newstring)
```

```python
#to get a string made of the 1st 2 and last 2 char
s=input("enter a string:")
newstring=s[1:2]+s[-2:]
print("new string:",newstring)
```

```python
#swapping
s1=input("enter first string:")
s2=input("enter second string:")
ns1=s2[0]+s1[1:]
ns2=s1[0]+s2[1:]
print("new string:",ns1,ns2)
```

```python
#replace()
s=input("enter a string:")
ns=s[0]+s[1:].replace(s[0],"$")
print("new string:",ns)
```
